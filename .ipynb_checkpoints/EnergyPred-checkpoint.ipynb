```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
%matplotlib inline
```

```python
#values range from 25695 to 9581. basically we need to predict how much energy will be consumed in future
```

```python
data['Datetime'] = pd.to_datetime(data['Datetime'])
data = data.set_index('Datetime')
```

```python

daily_pattern = data.groupby(data.index.hour)['AEP_MW'].mean()

plt.figure(figsize=(10,5))
daily_pattern.plot(kind='bar')
plt.title("Average Daily Energy Consumption Pattern")
plt.xlabel("Hour of Day (0-23)")
plt.ylabel("Average MW")
plt.show()
```

```python
#Now we need to elaborate the date to days, weeks, etc
```

```python
data['hour'] = data.index.hour
data['dayofweek'] = data.index.dayofweek
data['month'] = data.index.month
data['year'] = data.index.year
```

```python
data.head()
```

```python
#now, we generate lag features to give our model a memory of past events and for capturing temporal dependencies
```

```python
data['lag1'] = data['AEP_MW'].shift(1)
data['lag24'] = data['AEP_MW'].shift(24)
data['lag168'] = data['AEP_MW'].shift(168)
```

```python
data.head()
```

```python
#generating rolling features
```

```python
data['rolling_mean_24'] = data['AEP_MW'].rolling(24).mean()
data['rolling_std_24'] = data['AEP_MW'].rolling(24).std()

data['rolling_mean_168'] = data['AEP_MW'].rolling(168).mean()
```

```python
data.head()
```

```python
data.dropna(inplace=True)
```

```python
data.head()
```

```python
data.info()
```

```python
#now, for outliers, we check boxplots
```

```python
data.boxplot(column='AEP_MW', figsize=(5,5))
plt.show()
```

```python
#to resolve, we use iqr method
```

```python
Q1 = data['AEP_MW'].quantile(0.25)
Q3 = data['AEP_MW'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound: ", lower_bound)
print("Upper Bound: ", upper_bound)

outliers = data[(data['AEP_MW'] < lower_bound) | (data['AEP_MW'] > upper_bound)]
print("Number of Outliers: ", len(outliers))
```

```python
data.shape
```

```python
#we wont remove outliers since extreme values are possible in this case
```

```python
#in time series data, we dont random split data. we chronologically split the data.
#we would train our model on older(past) data and test on recent data to predict future better
```

```python
data = data.sort_values("Datetime")
```

```python
split_ratio = 0.8
split_point = int(len(data) * split_ratio)
```

```python
train = data.iloc[:split_point]
test = data.iloc[split_point:]
```

```python
train.shape
```

```python
test.shape
```

```python
train.head()
```

```python
test.head()
```

```python
target = 'AEP_MW'
feature_columns = ['hour', 'dayofweek', 'month',
                   'lag1', 'lag24', 'lag168',
                   'rolling_mean_24', 'rolling_std_24', 'rolling_mean_168']

X = data[feature_columns]
y = data[target]
```

```python
X_train, X_test = X.iloc[:split_point], X.iloc[split_point:]
y_train, y_test = y.iloc[:split_point], y.iloc[split_point:]
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
#First, we would use linear regression
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error,r2_score
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
y_pred_train = lr.predict(X_train)
y_pred_test = lr.predict(X_test)
```

```python
y_pred_train
```

```python
y_pred_test
```

```python
y_train.iloc[0]
```

```python
y_pred_train[0]
```

```python
train_rmse = np.sqrt(mean_squared_error(y_train, y_pred_train))
test_rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))

train_mae = mean_absolute_error(y_test, y_pred_test)
test_mae = mean_absolute_error(y_test, y_pred_test)
```

```python
r2 = r2_score(y_test, y_pred_test)
```

```python
print(train_rmse)
```

```python
print(test_rmse)
```

```python
print(train_mae)
```

```python
print(test_mae)
```

```python
print("R2 Score: ", r2)
```

```python
plt.figure(figsize=(12,6))
plt.plot(y_test.values, label="Actual", color='blue')
plt.plot(y_pred_test, label="Predicted", color='red')
plt.legend()
plt.title("Linear Regression - Actual vs Predicted Energy Consumption")
plt.show()
```

```python
#now, we use XGBOOST
```

```python
from xgboost import XGBRegressor
```

```python
xgb = XGBRegressor(
    n_estimators = 200,
    learning_rate = 0.1,
    max_depth = 6,
    random_state = 42
)
xgb.fit(X_train, y_train)
```

```python
xgb_y_pred = xgb.predict(X_test)

mse = mean_squared_error(y_test, xgb_y_pred)
r2 = r2_score(y_test, xgb_y_pred)
```

```python
print(mse)
```

```python
print("r2 score: ", r2)
```

```python
#THE R2 score of XGB is better than LinearRegression
```

```python
plt.figure(figsize=(8,5))
plt.barh(X.columns, xgb.feature_importances_)
plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.title("XGBoost Feature Importance")
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.scatter(y_test, xgb_y_pred, color='orange', alpha=0.6, edgecolors='k')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='blue', lw=2, linestyle='--')
plt.xlabel("Actual Energy Consumption")
plt.ylabel("Predicted Energy Consumption")
plt.title("Actual vs Predicted Energy Consumption")
plt.show()
```

```python
#Here we can see that many points lie close to the blue line(actual predictions) which make up to good predictions
```
